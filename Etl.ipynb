```python
import pandas as pd
```

```python
df = pd.read_csv("customers.csv")

print(df.head(20))    # First 5 rows
print(df.columns)  # Column names
print(df.shape)    # Number of rows and columns   
```

Standardize city

```python
df["city"] = df["city"].str.capitalize()
print(df.head(20))
```

Validate age until 18

```python
df = df[df["age"] >= 18]
print(df.head(20))
```

missing email

```python
df["email"] = df["email"].fillna('No email provided')
print(df.head(20))
```

remove duplicates

```python
df = df.drop_duplicates(subset=["email"], keep="first").drop_duplicates(subset=["customer_id"], keep="first")
print(df.head(20))
```

Fullname creation


```python
df["FullName"] = df["first_name"] + " " + df["last_name"]
print(df.head(20))
print(df.shape)
```

```python
df.to_csv("customers_clean.csv", index=False , header=True)
```

```python
df = pd.read_parquet("orders.parquet")

print(df.head(20))    # First 5 rows
print(df.columns)  # Column names
print(df.shape)    # Number of rows and columns
```

Remove duplicates

```python
df = df.drop_duplicates(subset=["order_id"], keep="first")
print(df.head(20))  
```

validate quantity

```python
df= df[df["quantity"] > 0]  
print(df.head(20))
```

convert order_date to dateTime

```python
df["order_date"] = pd.to_datetime(df["order_date"])
print(df.head(20))
print(df.dtypes)
```

validate customer id

```python
x = pd.read_csv("customers_clean.csv")

y = pd.merge(df, x, on="customer_id")
df = df[df["customer_id"].isin(y["customer_id"])]

print(df.head(20))
    
```

```python
df.to_parquet("orders_clean.parquet", index=False , engine="pyarrow")
```

```python
url = "https://raw.githubusercontent.com/MohammedHameds/test1455/refs/heads/main/products.csv"

df = pd.read_csv(url)
print(df.head())
```

Remove duplicates

```python
df=df.drop_duplicates(subset=["product_id"], keep="first").drop_duplicates(subset=["product_name"], keep="first")
print(df.head(20))       
```

standardize product/category

```python
df["product_name"] = df["product_name"].str.upper()
df["category"] = df["category"].str.upper()
print(df.head(20))
```

validate unit_price

```python
df=df[df["unit_price"] > 0]
print(df.head(20))
print(df.shape)
```

convert price to numeric

```python
df["unit_price"] = df["unit_price"].astype(float).round(2)
print(df.head(20))
print(df.dtypes)
```

```python
df.to_csv("products_clean.csv", index=False , header=True)
```
